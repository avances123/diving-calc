```python
import matplotlib.pyplot as plt
import numpy as np

from diving_calc.deco.tissue import Tissue, Tissues
```

```python
# Configuramos el entorno de plotting
#plt.style.use('seaborn')  # Estilo visualmente atractivo
plt.rcParams['figure.figsize'] = [10, 6]  # Tamaño de la figura


# Crear un objeto Tissue con los valores por defecto
todos = Tissues()
tissue = todos[0]
# Generar una serie de presiones ambiente entre 0 y 3 bar (presión atmosférica típica)
pressures = np.linspace(0, 10, 100)  # 100 puntos entre 0 y 3 bar
frontera = [0.79*p for p in pressures]

# Calcular los m-values correspondientes
m_values = [tissue.m_value(p) for p in pressures]

# Graficar
fig, ax = plt.subplots()
ax.plot(pressures, frontera, label="Frontera", linestyle="dashed", color="gray")
ax.plot(pressures,m_values)
ax.set_xlabel('Presión ambiente (bar)', fontsize=12)
ax.set_ylabel('Presión del tejido (bar)', fontsize=12)
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
plt.title('Relación entre la presión ambiente y la presion del tejido')


# Mostrar la gráfica
plt.show()

```

```python
# Configuramos el entorno de plotting
#plt.style.use('seaborn')  # Estilo visualmente atractivo
plt.rcParams['figure.figsize'] = [10, 6]  # Tamaño de la figura


# Crear un objeto Tissue con los valores por defecto
todos = Tissues()
# Generar una serie de presiones ambiente entre 0 y 3 bar (presión atmosférica típica)
pressures = np.linspace(0, 10, 100)  # 100 puntos entre 0 y 3 bar
frontera = [0.79*p for p in pressures]




# Graficar
fig, ax = plt.subplots()

ax.plot(pressures, frontera, label="Frontera", linestyle="dashed", color="gray")
for tissue in todos:
    # Calcular los m-values correspondientes
    m_values = [tissue.m_value(p) for p in pressures]
    ax.plot(pressures,m_values)

ax.set_xlabel('Presión ambiente (bar)', fontsize=12)
ax.set_ylabel('Presión del tejido (bar)', fontsize=12)
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
plt.title('Relación entre la presión ambiente y la presion del tejido')


# Mostrar la gráfica
plt.show()
```

```python
# Crear un objeto Tissues con los valores por defecto
todos = Tissues()

# Generar una serie de presiones ambiente entre 1 y 10 bar
pressures = np.linspace(1, 10, 100)
frontera = [p for p in pressures]

# Crear una figura con una cuadrícula de 4x4 subgráficas
fig, axes = plt.subplots(4, 4, figsize=(12, 12))  # 4 filas, 4 columnas

for i, (ax, tissue) in enumerate(zip(axes.flatten(), todos)):
    # Calcular los m-values correspondientes
    m_values = [tissue.m_value(p) for p in pressures]
    
    # Graficar la curva para este tejido en su subgráfico
    ax.plot(pressures, frontera, label="Frontera", linestyle="dashed", color="gray")
    ax.plot(pressures, m_values, label=f"Tejido {i+1}", color="b")
    
    # Configurar títulos y etiquetas
    ax.set_title(f"Tejido {i+1}")
    ax.set_xlim(1, 10)
    ax.set_ylim(1, 10)

# Ajustar los espacios entre subgráficas
plt.tight_layout()
plt.show()
```
