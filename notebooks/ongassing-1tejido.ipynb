```python
import matplotlib.pyplot as plt
import numpy as np

from diving_calc.deco.tissue import Tissue, Tissues
```

```python
presion_superficie = 1
presion_fondo = 4
x = Tissues()[9]
```

```python
def barra_presion(valor, limite_alerta):
    # Redondear valores a dos decimales
    valor = round(valor, 2)
    maximo = presion_fondo
    limite_alerta = round(limite_alerta, 2)

    fig, ax = plt.subplots(figsize=(6, 1.5))
    
    # Cambiar color si supera el límite de alerta
    color = 'royalblue' if valor <= limite_alerta else 'red'

    # Dibujar la barra de presión
    ax.barh(y=0, width=valor, height=0.5, color=color, align='center')

    # Dibujar el contorno de la barra
    ax.barh(y=0, width=maximo, height=0.5, color='none', edgecolor='black', linewidth=2)

    # Línea de alerta
    ax.axvline(x=limite_alerta, color='black', linestyle='--', linewidth=2)

    ax.set_xlim(0, maximo)
    ax.set_xticks([])
    ax.set_yticks([])
    
    # Texto dentro de la barra
    ax.text(valor / 2, 0, f"{valor} bar", va='center', ha='center', fontsize=12, color='white', fontweight='bold')

    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['bottom'].set_visible(False)
    ax.spines['left'].set_visible(False)

    # Agregar la leyenda arriba de la barra
    plt.text(0, 0.8, f"M-Value: {limite_alerta} bar", fontsize=12, color='black', fontweight='bold')

    plt.show()

# Ejemplo con presión de 120 bar, máximo 200 bar y límite de alerta en 150 bar



```

```python
presion_ambiente = presion_superficie # Estoy en la superficie
barra_presion(x.p_n2, x.m_value(presion_ambiente))
```

```python
presion_ambiente = presion_fondo # Bajo a 30m en 2 minutos

x.actualizar(1,2,presion_ambiente)
barra_presion(x.p_n2, x.m_value(presion_ambiente))
x.a
```

```python
presion_ambiente = presion_fondo # Buceo a 30m 40 minutos

x.actualizar(presion_ambiente,180,presion_ambiente)
barra_presion(x.p_n2, x.m_value(presion_ambiente))
```
